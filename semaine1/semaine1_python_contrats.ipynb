## Partie 1 — Les variables

Une **variable**, c'est une boîte avec un nom, dans laquelle tu ranges une information.
Le signe `=` veut dire « range ceci dans la boîte », pas « est égal à ».

Il existe plusieurs types d'informations :
- un **nombre entier** (`int`) : `5000`
- un **nombre à virgule** (`float`) : `0.245` — attention, en Python on écrit un **point**, pas une virgule
- un **texte** (`str`, pour *string*) : toujours entre guillemets, `"Amel Ben Salah"`
- un **vrai/faux** (`bool`) : `True` ou `False`

````python
nom_client = "Amel Ben Salah"
montant = 5000          # en dinars
duree_mois = 18
taux_nominal = 0.245    # 24,5 % s'écrit 0.245

print(nom_client)
print(montant)
print(type(montant), type(taux_nominal), type(nom_client))
````

Pour afficher un texte qui mélange des mots et des variables, on utilise une **f-string** : un `f` devant les guillemets, et les variables entre accolades `{ }`.

````python
print(f"{nom_client} emprunte {montant} DT sur {duree_mois} mois.")

# On peut faire des calculs directement
mensualite_simple = montant / duree_mois
print(f"Mensualité hors intérêts : {mensualite_simple} DT")

# :.2f veut dire « avec 2 chiffres après la virgule »
print(f"Mensualité hors intérêts : {mensualite_simple:.2f} DT")
````

## Partie 3 — Les dictionnaires

Un **dictionnaire** range des informations avec une **étiquette** (la *clé*) devant chacune. On l'écrit entre accolades `{ }`, avec `clé: valeur`.

C'est exactement la forme d'un contrat : chaque information a un nom. **Tout ton projet va manipuler des dictionnaires.**

````python
contrat = {
    "numero": "SYN-0001",
    "nom": "Amel Ben Salah",
    "cin": "08123456",          # un CIN est un texte, pas un nombre : on garde le 0 du début
    "montant": 5000,
    "montant_lettres": "cinq mille dinars",
    "duree_mois": 18,
    "taux_nominal": 0.245,
    "teg": 0.262,
    "garantie": "caution solidaire",
}

print(contrat["nom"])                   # lire une valeur avec sa clé
print(contrat["montant"])
````

````python
contrat["duree_mois"] = 24              # modifier une valeur
contrat["assurance"] = "oui"            # ajouter une nouvelle clé
print(contrat)

print(contrat.keys())                   # toutes les clés
print("teg" in contrat)                 # « in » vérifie si une CLÉ existe
````

**Piège fréquent :** si tu demandes une clé qui n'existe pas, Python plante (`KeyError`). Pour éviter ça, utilise `.get()` : il renvoie `None` (« rien ») au lieu de planter. Très utile pour détecter une mention **absente**.

````python
print(contrat.get("frais"))             # None : la clé n'existe pas
print(contrat.get("frais", "ABSENT"))   # on peut choisir ce qu'il renvoie
````

## Partie 4 — Les conditions

Une **condition** dit : « **si** ceci est vrai, **alors** fais cela, **sinon** fais autre chose ».

⚠️ En Python, ce qui est « à l'intérieur » du `if` est **décalé de 4 espaces** (l'indentation). Ce n'est pas décoratif : c'est comme ça que Python sait ce qui dépend de la condition.

Exemple : d'après l'**arrêté du 13 avril 2018** (qui modifie celui du 18 janvier 2012), une institution de microfinance constituée en société anonyme — c'est le cas de Microcred SA — ne peut pas prêter plus de **40 000 DT**, sur **7 ans maximum** (à vérifier dans le texte).

````python
PLAFOND_SA = 40000          # une constante : on l'écrit en MAJUSCULES par convention
DUREE_MAX_MOIS = 7 * 12

montant = 45000

if montant > PLAFOND_SA:
    print(f"ANOMALIE : {montant} DT dépasse le plafond de {PLAFOND_SA} DT")
else:
    print("Montant conforme au plafond")
````

Les **comparaisons** : `==` (égal — deux signes !), `!=` (différent), `>`, `<`, `>=`, `<=`.

On peut combiner : `and` (les deux doivent être vrais), `or` (au moins un), `not` (l'inverse).

````python
cin = "0812345"

# Un CIN tunisien : 8 caractères, uniquement des chiffres
if len(cin) == 8 and cin.isdigit():
    print("CIN valide")
else:
    print(f"ANOMALIE : CIN '{cin}' invalide ({len(cin)} caractères)")
````

````python
# elif = « sinon, si » : pour plusieurs cas
teg = 0.31
seuil = 0.30          # valeur inventée pour l'exemple : le vrai seuil est publié par l'ACM

if teg > seuil:
    print("BLOQUANT : TEG au-dessus du seuil")
elif teg > seuil - 0.01:
    print("À VÉRIFIER : TEG très proche du seuil")
else:
    print("TEG conforme")
````

## Partie 5 — Les boucles

Une **boucle `for`** répète une action **pour chaque élément** d'une liste. C'est là que l'automatisation commence : au lieu de vérifier une mention, tu les vérifies **toutes**.

````python
contrat = {
    "montant": 5000,
    "taux_nominal": 0.245,
    "frais": 50,
    "modalites_remboursement": "18 mensualités",
    # pas de "teg", pas de "cout_global", pas de "assurance"
}

cles_obligatoires = ["montant", "taux_nominal", "frais", "cout_global",
                     "teg", "modalites_remboursement", "assurance"]

for cle in cles_obligatoires:
    if cle in contrat:
        print(f"  ok      {cle}")
    else:
        print(f"  MANQUE  {cle}")
````

On peut aussi **remplir une liste** pendant la boucle — c'est exactement ce que fera ton vérificateur : collecter les anomalies.

````python
manques = []
for cle in cles_obligatoires:
    if cle not in contrat:
        manques.append(cle)

print(f"{len(manques)} mention(s) manquante(s) : {manques}")
````

````python
# Parcourir un dictionnaire : .items() donne la clé ET la valeur
for cle, valeur in contrat.items():
    print(f"{cle} -> {valeur}")
````

## Partie 6 — Les fonctions

Une **fonction**, c'est un bloc de code auquel tu donnes un **nom**, que tu peux réutiliser autant de fois que tu veux.
- `def` pour la définir
- entre parenthèses : ce qu'elle **reçoit** (les *paramètres*)
- `return` : ce qu'elle **renvoie**

Dans ton projet, la règle d'or sera : **une règle = une fonction**. Ajouter une règle = ajouter une fonction, sans rien casser.

````python
def verifier_plafond(contrat):
    """Renvoie un message d'anomalie si le montant dépasse le plafond, sinon None."""
    if contrat["montant"] > PLAFOND_SA:
        return f"Montant {contrat['montant']} DT > plafond {PLAFOND_SA} DT"
    return None

print(verifier_plafond({"montant": 5000}))
print(verifier_plafond({"montant": 50000}))
````

````python
def verifier_mentions(contrat, cles_obligatoires):
    """Renvoie la liste des mentions obligatoires absentes du contrat."""
    manques = []
    for cle in cles_obligatoires:
        if cle not in contrat:
            manques.append(f"Mention absente : {cle}")
    return manques

print(verifier_mentions({"montant": 5000, "teg": 0.26}, ["montant", "teg", "frais"]))
````

### Solution 7 — le mini-vérificateur

````python
def verifier_duree(contrat):
    if contrat.get("duree_mois", 0) > DUREE_MAX_MOIS:
        return f"Durée {contrat['duree_mois']} mois > maximum {DUREE_MAX_MOIS} mois"
    return None

def verifier_contrat(contrat):
    anomalies = []

    # Règle 1 : les mentions obligatoires (renvoie une LISTE -> on l'ajoute en entier)
    anomalies.extend(verifier_mentions(contrat, cles_obligatoires))

    # Règles 2 à 4 : chacune renvoie un message ou None
    resultats = []
    if "montant" in contrat:
        resultats.append(verifier_plafond(contrat))
    resultats.append(verifier_cin(contrat))
    resultats.append(verifier_duree(contrat))

    for r in resultats:
        if r is not None:
            anomalies.append(r)
    return anomalies

for c in corpus:
    anomalies = verifier_contrat(c)
    if len(anomalies) == 0:
        print(f"{c['numero']} : conforme")
    else:
        print(f"{c['numero']} : {len(anomalies)} anomalie(s)")
        for a in anomalies:
            print(f"   - {a}")
````
